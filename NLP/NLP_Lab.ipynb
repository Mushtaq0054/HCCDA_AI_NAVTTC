### **Removing Punctuation**

```python
import string

# Import the string module, which contains a list of punctuation characters

def remove_punctuation(text):
    # Create a translation table that maps each punctuation character to None
    #translator = str.maketrans('PAKISTANI','*********', string.punctuation)
    translator = str.maketrans('or', 'ro', string.punctuation)
    # Remove punctuation from the text using the translation table
    return text.translate(translator)

# Example usage
# sample_text = "Hello, World! This is NLP preprocessing - removing punctuation."
sample_text = "Hello guys here myself Mushtaq Ahmad in front of you "
clean_text = remove_punctuation(sample_text)

# Print the original text
print("Original text:", sample_text)
# Print the text after removing punctuation
print("Text without punctuation:", clean_text)
```

### **Removing Numbers**

```python
import re
# Import the regular expressions module

# Define a function to remove numbers from a given text
def remove_numbers(text):
    # Use re.sub to replace all digit sequences with an empty string
    return re.sub(r'\d+', ' "Representaion of No" ', text)

# Example usage
sample_text = "In 2023, ChatGPT had over 100 million users."  # Define a sample text containing numbers
clean_text = remove_numbers(sample_text)  # Remove numbers from the sample text

print("Original text:", sample_text)  # Print the original text
print("Text without numbers:", clean_text)  # Print the text after removing numbers
```

```python
import re

# Import the regular expressions module

# Define a text string containing multiple sentences
text_1 = "Q: What's this? A: It's a test!"
text_2 = "Hello! How are you = I'm fine, thank you."

# Split the text into sentences using a regular expression
# The pattern splits at punctuation (.,!,?) followed by whitespace and a capital letter
sentences_1 = re.split(r'(?<=[.!?])\s+(?=[A-Z])', text_1)
sentences_2 = re.split(r'(?<=[.!?])\s+(?=[A-Z])', text_2)

# Print the resulting list of sentences
print(text_1)
print(text_2)
print(sentences_1)
print(sentences_2)
```

### **Porter Stemmer (Most Popular)**

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["running", "happiness", "flies", "unhappy"]

stemmed = [stemmer.stem(word) for word in words]
print(stemmed)
```

```python
import re

pattern = r'P\d{3}_\w{2,}'
test_strings = [
    "Here is a valid code: P123_ab",
    "Wrong code: P12_ab",
    "Another good one: P999_xyz123",
    "Missing underscore: P123ab",
    "Too short: P123_a"
]

print("Matching product codes (P123_ab format):\n")
for text in test_strings:
    match = re.search(pattern, text)
    if match:
        print(f"✔ Found: {match.group()} in \"{text}\"")
    else:
        print(f"✘ No match in \"{text}\"")
```

### **1. re.search()**

```python
import re

text = "The rain in Spain falls mainly in the plain."

# Search for the first word starting with 'S'
result = re.search(r"\bS\w+", text)  

if result:
    print("Found:", result.group())  # Output: "Spain"
    print("Start index:", result.start())  # 12
    print("End index:", result.end())    # 17
else:
    print("No match")
```

### **2. re.match()**

```python
text = "42 is the answer."

# Match only if the string starts with a number
result = re.match(r"\d+", text)  

if result:
    print("Starts with number:", result.group())  # Output: "42"
else:
    print("No match at start")

# Fails if text starts with non-digit:
re.match(r"\d+", "The answer is 42")  # Returns None
```

### **3. re.findall()**

```python
text = "Emails: user1@test.com, user2@example.com"

# Find all email addresses
emails = re.findall(r"\b[\w.-]+@[\w.-]+\.\w+\b", text)  

print(emails)  
# Output: ['user1@test.com', 'user2@example.com']
```

### **4. re.sub()**

```python
text = "Order #12345 shipped on 12/10/2023"

# Replace all numbers with '[NUM]'
new_text = re.sub(r"\d+", "[NUM]", text)  

print(new_text)  
# Output: "Order #[NUM] shipped on [NUM]/[NUM]/[NUM]"
```
