# cascade download and load  by online

```python
# cascade download and load
import cv2
import os
import urllib.request

# Helper function to ensure Haar Cascade XMLs are downloaded locally
def ensure_cascade(filename, url):
    if not os.path.exists(filename) or os.path.getsize(filename) < 1000:
        print(f"📥 Downloading {filename}...")
        urllib.request.urlretrieve(url, filename)
        print(f"✅ Saved {filename}")
    return filename

# Ensure XML files are available
FACE_CASCADE_PATH = ensure_cascade(
    "haarcascade_frontalface_default.xml",
    "https://raw.githubusercontent.com/opencv/opencv/master/data/haarcascades/haarcascade_frontalface_default.xml"
)
EYE_CASCADE_PATH = ensure_cascade(
    "haarcascade_eye.xml",
    "https://raw.githubusercontent.com/opencv/opencv/master/data/haarcascades/haarcascade_eye.xml"
)

# Load the pre-trained models into memory
face_cascade = cv2.CascadeClassifier(FACE_CASCADE_PATH)
eye_cascade = cv2.CascadeClassifier(EYE_CASCADE_PATH)

# Verify if models loaded properly
if face_cascade.empty():
    raise RuntimeError("Failed to load face cascade classifier!")
else:
    print("🎉 Haar Cascades loaded successfully and ready to use!")
```

# only detect only one face

```python
# only detect only one face
# Initialize webcam (0 for default camera)
cap = cv2.VideoCapture(0)

print("📷 Webcam stream started. Look into the camera! Press 'q' to exit.")

while True:
    # Read frame from camera
    ret, img = cap.read()
    if not ret or img is None:
        print("Error: Could not read frame from webcam.")
        break

    # Convert to grayscale (face detection works on grayscale)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # Draw rectangles around detected faces
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)  # Blue rectangle
        cv2.putText(img, 'Face', (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (255, 0, 0), 2)

    # Display output
    cv2.imshow('Face Detection', img)

    # Exit on 'q' key press
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources cleanly
cap.release()
cv2.destroyAllWindows()
print("🛑 Webcam stream closed.")
```

# Detect multiple faces with unique labels

```python
# Detect multiple faces with unique labels
# Load the pre-trained Haar Cascade face detector
# Make sure you have initialized the 'face_cascade' before running this loop
# face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
# face_cascade = cv2.CascadeClassifier(FACE_CASCADE_PATH)
# Initialize webcam (0 for default camera)
cap = cv2.VideoCapture(0)

print("📷 Webcam stream started. Look into the camera! Press 'q' to exit.")

while True:
    # Read frame from camera
    ret, img = cap.read()
    if not ret or img is None:
        print("Error: Could not read frame from webcam.")
        break

    # Convert to grayscale (face detection works on grayscale)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect all faces in the frame
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    # Initialize a counter for counting faces in the current frame
    face_count = 0

    # Draw rectangles and unique labels around all detected faces
    for (x, y, w, h) in faces:
        face_count += 1  # Increment the count for each face found
        
        # Draw a blue rectangle around the detected face
        cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)  
        
        # Create a dynamic label (e.g., 'Face 1', 'Face 2', etc.)
        label = f"Face {face_count}"
        
        # Put the dynamic label above the rectangle
        cv2.putText(img, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (255, 0, 0), 2)

    # Display output window
    cv2.imshow('Multi-Face Detection', img)

    # Exit on 'q' key press
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources cleanly
cap.release()
cv2.destroyAllWindows()
print("🛑 Webcam stream closed.")
```
