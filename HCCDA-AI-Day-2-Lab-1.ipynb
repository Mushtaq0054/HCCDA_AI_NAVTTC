##**TUPLE** <br>
tupple is a immutable <br>
can't cahnge any Element in tupple <br>
tuple use small brackets

```python
# Make a standard tupple
fruits = ("apple", "banana", "cherry")
print("Original Tuple:", fruits)
```

```python
fruits = ("apple", "banana", "cherry")
# Mistake: try to change index 1 (banana) to "mango"
# rememmber it will no change because tupple is immutable data
fruits[1] = "mango"

# ❌ Error: TypeError: 'tuple' object does not support item assignment
# The error shows in line no 4
```

What We CAN Do with a Tuple **(Allowed Operations)**
we cannot change elements, but you can read ** (access) ** them and use **slicing** just like a list:


```python

numbers = (10, 20, 30, 40, 50)

# A. Accessing a specific element
print(numbers[2])   # Prints 30 (the element at index 2)

# B. Slicing (Extracting a part of the tuple)
print(numbers[1:4]) # Prints (20, 30, 40) (index 4 is excluded)

# C. Checking total length
print(len(numbers)) # Prints 5 (total number of elements)
```

**Special Tip:** Single Element Tuple <br>
If you want to create a tuple with only ONE element, you MUST add a trailing comma `,`. <br>
Otherwise, Python will treat it as a regular string or integer.


```python
wrong_tuple = ("apple")  # This creates a standard String, NOT a tuple
right_tuple = ("apple",) # Adding the comma makes it a valid Tuple

# Display
print(wrong_tuple)
print(right_tuple)
```

```python
# Display their types
print(type(wrong_tuple)) # This will show string datatype
print(type(right_tuple)) # This will show tuple datatype
```

# Dictionaries

*   **Definition:** A Python Dictionary (dict) stores data in **key: value pairs** inside curly brackets {}.
*   **No Index Numbers:** Unlike lists, you do not look up items using numbers like 0 or 1. Instead, you use a unique **Key** to access its **Value**.
*   Dictionaries are **mutable**.
*  Keys must be unique (no duplicates allowed).


```python
student = {
    "name": "Ahmad",
    "age": 22,
    "course": "Python",
    "grade": "A"
}

print(student)
```

```python
# 2. ACCESSING VALUES
print(student["name"])      # Print only students name
print(student["course"])     # Print only course
```

```python
# Adding a  new key-value pair
student["city"] = "Lahore"  # Adds "city": "Lahore" to the dictionary
print(student)
```

```python
student["Remarks"] = "Good"
print(student)
```

**NUMPY**  
<br>
*   **NumPy** stands for Numerical Python. It is a powerful library used for working with numerical data in the form of arrays.
*  **array** is a collection of elements of the same data type.
*   **Dimensions:**
    *   **1D Array (1-Dimensional):** A single row of elements (like a vector).
    *   **2D Array (2-Dimensional):** Elements arranged in rows and columns (like a matrix or table).


```python
import numpy as np

# 1D ARRAY (Single row)
Dimension_1 = np.array([1, 2, 5])

# 2. 2D ARRAY (Rows and Columns)
# Note: It contains a list of lists inside square brackets [[...], [...]]
Dimension_2 = np.array([[1, 5, 7], [3, 95, 6]])
```

```python
print(Dimension_1) # Display 1 D array

print(Dimension_2) # Display 2 D array
```
