# List

```python
# List concatenation is like string concatenation
a = [1,2,3,4]
b = [5,3,7,8]

print("list concatenation a + b", a + b)
c = [1,3.4,"hello"]
d = [4,5,6]
print("list concatenation c + d", c + d)

 # Element-wise concatenation
x = [[1,2], [3,4]]
y = [[5,1], [7,8]]
print("list concatenation x + y", x + y)

# list and tupple concatenation
a = [1,2,3,4]
b = (5,3,7,8) 

print("list concatenation a + list(b)", a + list(b)) # we can convert tupple into list and then concatenate it with list
# we can not concatenate list and tupple, it will give us error
print("list concatenation a + b", a + b) # it will give us error
```
