### Why numpy?

```python
def sum_python(n):
    a = [i for i in range(n)]
    b = [i for i in range(n)]
    c = [None] * n
    start = time.perf_counter()
    for i in range(n):
        c[i] = a[i] + b[i]
    end = time.perf_counter()
    return end - start

def sum_numpy(n):
    a = np.arange(n)
    b = np.arange(n)
    start = time.perf_counter()
    c = a + b
    end = time.perf_counter()
    return end - start

experiments = [100, 1000, 10000, 1000000]
python_results = [sum_python(n) for n in experiments]
numpy_results = [sum_numpy(n) for n in experiments]

plt.figure(figsize=(8, 6))
plt.xlabel("Array size", size=14)
plt.ylabel("Execution time", size=14)
plt.title("Execution time in seconds", size=18)
sns.lineplot(x=experiments, y=python_results, color='r', label='Python list')
sns.lineplot(x=experiments, y=numpy_results, color='b', label='NumPy')
```

### Applying a mathematical formula: $f(x) = \frac {1} {\sigma \sqrt{2\pi}} e^{-\frac {1}{2} \left(\frac {x - \mu} {\sigma}\right)^2}$

```python
mu = 0
std = 1
x = np.linspace(-5, 5, 100)
y = 1 / (std * np.sqrt(2 * np.pi)) * np.exp(-1/2 * np.square((x - mu)/std))

sns.lineplot(x=x, y=y)
plt.title("Standard normal distribution")
```

###### Single value decomposition (SVD)

```python
china = load_sample_image('china.jpg')
```

```python
china.shape
```

```python
china.min(), china.max()
```

```python
china = china / 255.
```

```python
china.min(), china.max()
```

```python
plt.imshow(china)
```

```python
china = np.transpose(china, (2, 0, 1))
```

```python
china.shape
```

```python
plt.imshow(china[0], cmap="Reds")
```

```python
plt.imshow(china[1], cmap="Greens")
```

```python
plt.imshow(china[2], cmap="Blues")
```

```python
k = 30
```

```python
U, s, Vt = np.linalg.svd(china)
U.shape, s.shape, Vt.shape
```

```python
s = np.apply_along_axis(np.diagflat, 1, s)
s.shape
```

```python
low_rank_china = U @ s[..., :k] @ Vt[..., :k, :]
```

```python
low_rank_china = np.transpose(low_rank_china, (1,2,0))
```

```python
plt.imshow(low_rank_china)
```

### Plotting the Julia set

```python
c = complex(-0.835, -0.2321)
```

```python
x = np.linspace(-2, 2, 2000)
y = np.linspace(-2, 2, 2000)
(XX, YY) = np.meshgrid(x,y)
XX.shape, YY.shape
```

```python
Z = np.array(XX + 1j * YY)
result = np.zeros(Z.shape)
mask = np.ones(Z.shape, dtype=bool)
for i in range(24):
    Z[mask] = Z[mask] ** 2 + c
    mask = np.abs(Z) < 2
    result += mask
```

```python
plt.figure(figsize=(12,12))
plt.imshow(result, extent=(x.min(), x.max(), y.min(), y.max()), cmap=cm.gnuplot2)
plt.axis("off")
plt.tight_layout()
plt.show()
```
