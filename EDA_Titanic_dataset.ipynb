# 6 Data Cleaning & Imputation

```python
# Remove null values from the dataset
df_cleaned = df.dropna() # drop rows with any null values
df_cleaned
df_cleaned.isnull().sum() # check for missing values in the cleaned dataset
```

### After removing the missing value

```python
# Specific Numerical Measures 
print("Key Numerical Metrics (Age & Fare):")
print("Mean Age:", df['age'].mean())
print("Median Age (50%):", df['age'].median())
print("Min Age:", df['age'].min())
print("Max Age:", df['age'].max())
print("Min Fare:", df['fare'].min())
print("Max Fare:", df['fare'].max())
```

```python
# 2. Numerical Columns ko Impute Karna (Fill Missing Values)
# Agar skewness kam ho toh mean, agar outliers zyada hon toh median use karte hain
df['age'] = df['age'].fillna(df['age'].median()) # Ya df['age'].mean()
df['fare'] = df['fare'].fillna(df['fare'].median()) #

# 3. Categorical Columns ko Impute Karna
# Mode se bharna (Sab se common category)
embarked = df['embarked'].fillna(df['embarked'].mode()[0]) # this will fill the missing values in the 'embarked' column with the most frequent value (mode) of that column.
sex = df['sex'].fillna(df['sex'].mode()[0]) # this will fill the missing values in the 'sex' column with the most frequent value (mode) of that column.
who = df['who'].fillna(df['who'].mode()[0]) # this will fill the missing values in the 'who' column with the most frequent value (mode) of that column.
deck = df['deck'].fillna(df['deck'].mode()[0]) # this will fill the missing values in the 'deck' column with the most frequent value (mode) of that column.
embark_town = df['embark_town'].fillna(df['embark_town'].mode()[0]) # this will fill the missing values in the 'embark_town' column with the most frequent value (mode) of that column.
alive = df['alive'].fillna(df['alive'].mode()[0]) # this will fill the missing values in the 'alive' column with the most frequent value (mode) of that column.
alone = df['alone'].fillna(df['alone'].mode()[0]) # this will fill the missing values in the 'alone' column with the most frequent value (mode) of that column.
```

# Data Visualisation

```python
df.plot(kind='scatter', x='age', y='fare', color='blue')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
# 1. Sabse badi umar wale musafir ka pura record
oldest = df.loc[df['age'].idxmax()]

print("--- Sabse Badi Umar Ka Musafir ---")
print("Umar (Age):", oldest['age'])
print("Gender:", oldest['sex'])
print("Category:", oldest['who'])
print("Class:", oldest['class'])
print("Kiraya (Fare):", oldest['fare'])
print("Kya Zinda Bacha?:", oldest['alive'])
```

```python
# Top 5 sabse badi umar ke log
df.sort_values(by='age', ascending=False)[['age', 'sex', 'class', 'who', 'alive']].head(5)
```

```python
# 2. Sabse kam umar wale musafir ka record
youngest = df.loc[df['age'].idxmin()]

print("--- Sabse Chhota Musafir ---")
print("Umar (Age):", youngest['age'])  # Decimal mein aayega (e.g. 0.42 saal)
print("Gender:", youngest['sex'])
print("Class:", youngest['class'])
print("Kya Zinda Bacha?:", youngest['alive'])
```

```python
df.sort_values(by='age', ascending=True)[['age', 'sex', 'class', 'who', 'alive']].head(5)
```

```python
# 3. Maximum fare nikal kar un logon ka data dekhna
max_fare_value = df['fare'].max()
highest_fare_passengers = df[df['fare'] == max_fare_value]

print("--- Sabse Zyada Kiraya Dene Wale ---")
print("Max Kiraya Kitna Tha:", max_fare_value)
display(highest_fare_passengers[['sex', 'age', 'class', 'who', 'fare', 'embark_town', 'alive']])
```

```python
# 4. Class-wise Fare Summary
class_fare_summary = df.groupby('class')['fare'].agg(
    Tadaad='count',
    Kam_se_Kam='min',
    Average='mean',
    Darmiyana_Median='median',
    Sabse_Zyada='max'
)

display(class_fare_summary)
```

```python
# 5a. Sirf Gender ki tadaad (Male vs Female)
print("--- Male vs Female Count ---")
print(df['sex'].value_counts())

# 5b. Man, Woman aur Child ki alag alag tadaad
print("\n--- Men, Women aur Children Count ---")
print(df['who'].value_counts())

# 5c. Har category ka percentage (%)
print("\n--- Category Percentage ---")
print((df['who'].value_counts(normalize=True) * 100).round(2))
```

```python
# 6a. Men, Women aur Children mein se kitne bache aur kitne mare
print("--- Who vs Alive (Count) ---")
print(pd.crosstab(df['who'], df['alive']))

# 6b. Bachne ka percentage (Survival Rate) har category ka
print("\n--- Who vs Survival Rate (%) ---")
print((df.groupby('who')['survived'].mean() * 100).round(2))

# 6c. Gender + Class ka aapas mein combined survival rate
print("\n--- Gender aur Class ka Combined Survival Rate (%) ---")
print((pd.pivot_table(df, values='survived', index='sex', columns='class', aggfunc='mean') * 100).round(2))
```
