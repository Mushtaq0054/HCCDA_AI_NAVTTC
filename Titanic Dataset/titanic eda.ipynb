# Binnings

```python
Bins = [0, 2, 5 ,12,18, 30,50,80]
Labels = [ 'Infant', 'Toddler', 'Child', 'Teenager', 'Young Adult', 'Adult', 'old']
kashti['age-group']=pd.cut(kashti['age'], bins=Bins, labels=Labels)
print(kashti['age-group'])

```

```python
print(kashti[['age', 'age-group']].head(10))
```

```python
kashti.head()
```

# Group by new age categories and calculate survival rate

```python
# Group by new age categories and calculate survival rate
kashti['survival-group-by-age'] = kashti.groupby('age-group')['survived'].transform('mean')
print(kashti['survival-group-by-age'])
```

```python
kashti['survival-group-by-age'] = kashti.groupby('age-group')['survived'].transform('mean')
print(kashti[['age', 'age-group', 'survival-group-by-age']].head(10))
```

# Group by new age categories and calculate class type  .mean in which class thy come

```python
# Group by new age categories and calculate class type  .mean in which class thy come
# Calculate the most common class for each age group and assign to each row
kashti['class-type-by-age'] = kashti.groupby('age-group')['class'].transform(lambda x: x.mode()[0])

# Display the first 10 rows to check
print(kashti[['age', 'age-group', 'class-type-by-age','survival-group-by-age']].head(10))
```

# Data Filtering

```python
# Find all the female passengers
FEMALES = kashti[kashti['sex']=='female']
print(FEMALES)
```

```python
# kashti['sex'] == 'female' → creates a Boolean Series (True for females, False for males).
# This Boolean Series is stored in a new column female_passengers inside the same kashti DataFrame.
# is code ma hm aik new column create karty hain
kashti['female_passengers'] = kashti['sex'] == 'female'
print(kashti[['sex', 'female_passengers']].head(10))
```

```python
# kashti[kashti['sex'] == 'female'] → filters the dataset to only include rows where sex is female.
# It creates a new DataFrame female_passengers that contains only female passengers.
# is ma hm aik new dataframe create karty hain

female_passengers = kashti[kashti['sex'] == 'female']
print(female_passengers.head(10))
```
