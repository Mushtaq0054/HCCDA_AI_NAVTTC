# Data Visualiation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
import plotly.io as pio
```

# Load Dataset

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

How to plot graph

```python
df.plot(kind='scatter', x='age', y='fare', color='blue')
```

```python
df.plot(kind='scatter', x='age', y='fare', alpha=0.5, color="#F2FA04")
```

how to lable x-axis and y-axis

```python
df.plot(kind='scatter', x='age', y='fare', color='blue')

plt.xlabel("Age of Passenger")
plt.ylabel("Fare of Passenger")
```

# Make plots by using Plotly

```python
fig = px.scatter(df, x='age', y='fare', color='class')
```

```python
# display figure
fig.show()
```

```python
fig = px.scatter(df, x='age', y='fare', color='who')
fig = plt.xlabel='age of passenger'
fig = plt.ylabel='fare of passenger'
```

```python
fig = px.scatter(df, 
                 x='age', 
                 y='fare', 
                 color='who',
                 labels={'age': 'age of passenger', 
                         'fare': 'fare of passenger'})
fig.show()
```
