```python
# =====================================================
# 🚢 Exploratory Data Analysis (EDA) on Titanic Dataset
# =====================================================

# 🔹 Step 1: Import Libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

# Display settings
pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
```

```python

# 🔹 Load Titanic dataset
kashti = sns.load_dataset('titanic')

# =====================================================
# 🔹 Step 2: Basic Data Understanding
# =====================================================
print("📌 First 5 Rows of Dataset:")
display(kashti.head())

print("\n📌 Last 5 Rows of Dataset:")
display(kashti.tail())

print("\n📌 Dataset Shape (Rows, Columns):", kashti.shape)
print("\n📌 Column Names:", kashti.columns.tolist())

print("\n📌 Dataset Info:")
kashti.info()

print("\n📌 Statistical Summary (Numerical Columns):")
display(kashti.describe())

print("\n📌 Statistical Summary (Categorical Columns):")
display(kashti.describe(include='object')) # it show just categorical columns
# display(kashti.describe(include='all')) # it show all columns
# display(kashti.describe(include='number')) # it show just numerical columns
# display(kashti.describe(include='bool')) # it show just boolean columns
# display(kashti.describe(include='datetime')) # it show just datetime columns
# display(kashti.describe(include='category')) # it show just category columns
# display(kashti.describe(include='float')) # it show just float columns
# display(kashti.describe(include='int')) # it show just int columns
# display(kashti.describe(include='string')) # it show just string columns
# display(kashti.describe(include='object', datetime_is_numeric=True)) # it show just object columns with datetime as numeric
# display(kashti.describe(include='all', datetime_is_numeric=True)) # it show all columns with datetime as numeric
```

```python

# =====================================================
# 🔹 Step 3: Missing Values & Data Cleaning
# =====================================================
print("\n📌 Missing Values Before Cleaning:")
print(kashti.isnull().sum())

# Fill missing numeric values with mean
kashti['age'] = kashti['age'].fillna(kashti['age'].mean())
kashti['fare'] = kashti['fare'].fillna(kashti['fare'].mean())

# Fill missing categorical values with mode or 'Unknown'
kashti['embarked'] = kashti['embarked'].fillna(kashti['embarked'].mode()[0])
kashti['embark_town'] = kashti['embark_town'].fillna('Unknown')
kashti['deck'] = kashti['deck'].fillna('Unknown')

print("\n📌 Missing Values After Cleaning:")
print(kashti.isnull().sum())
```

```python

# =====================================================
# 🔹 Step 4: Univariate Analysis
# =====================================================
plt.figure(figsize=(6,4))
sns.countplot(x='survived', data=kashti, palette='Set2')
plt.title("Survival Count")
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(kashti['age'], kde=True, bins=30, color='skyblue')
plt.title("Age Distribution")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(y='fare', data=kashti, palette='Pastel1')
plt.title("Fare Outliers")
plt.show()
```

```python

# =====================================================
# 🔹 Step 5: Bivariate Analysis
# =====================================================
plt.figure(figsize=(7,4))
sns.countplot(x='sex', hue='survived', data=kashti, palette='Set1')
plt.title("Survival by Gender")
plt.show()

plt.figure(figsize=(7,4))
sns.countplot(x='class', hue='survived', data=kashti, palette='Set3')
plt.title("Survival by Passenger Class")
plt.show()

plt.figure(figsize=(7,5))
sns.scatterplot(x='age', y='fare', hue='survived', style='sex', data=kashti, palette='coolwarm')
plt.title("Age vs Fare (Survival)")
plt.show()
```

```python

# =====================================================
# 🔹 Step 6: Multivariate Analysis
# =====================================================
plt.figure(figsize=(8,5))
sns.heatmap(kashti.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

# Pairplot (can take time)
sns.pairplot(kashti, hue='survived', diag_kind='hist')
plt.show()
```

```python

# =====================================================
# 🔹 Step 7: Additional Insights
# =====================================================
print("\n📌 Survival Rate by Passenger Class:")
print(kashti.groupby('class')['survived'].mean())

pivot = pd.pivot_table(kashti, values='survived', index='sex', columns='class', aggfunc='mean')
print("\n📌 Survival Rate by Gender & Class:")
print(pivot)

plt.figure(figsize=(5,4))
sns.heatmap(pivot, annot=True, cmap='YlGnBu', linewidths=1)
plt.title("Survival Rate by Gender & Class")
plt.show()
```
