# Numpy 2D Array
**2D array in numpy is a matrix, and it has rows and columns, so we can access the elements of the matrix by using the index of the row and column, and the index start from 0, so d2[0,0] means row 0 and column 0, which is 1**

```python
d2 = np.array([
    [1,2],
    [10,20], 
    [1,3],
    [100,200],
    [1000,2000]
    ])

print("2D array: \n",d2)
print("\n Number of dimensions:",d2.ndim)  
print("\n Shape of array:",d2.shape)
print("\n Number of rows:",d2.shape[0]) 
print("\n Number of columns:",d2.shape[1])
# how many elements are there in the array (row * col)
print("\n size of array:",d2.size)
```

# Slicing

```python
# Slicing is a technique to extract a portion of an array. In 2D arrays, slicing can be done using the following syntax: array[row,column]
# make a 2d array of 5 rows and 2 columns and slice it to get the first 3 rows and the first column.
d2 = np.array([
    [1,2],
    [10,20], 
    [1,3],
    [100,200],
    [1000,2000]
    ])

print("2D array: \n",d2)
print("\n Sliced 2D array: \n",d2[:3, :1])
```
